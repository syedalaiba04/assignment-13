```python
import pandas as pd
```

## 1. Load the cleaned teaching file

```python
df = pd.read_csv("teaching_prices_clean (1).csv")

print("Dataset loaded successfully.")
print("Number of rows and columns:", df.shape)

display(df.head())
```

## 2. Calculate the mean and median

I calculated the mean and median separately for each item. Missing prices are automatically left out of these calculations.

```python
print("Staple A")
print("Mean:", df[df["item"] == "Staple A"]["price"].mean())
print("Median:", df[df["item"] == "Staple A"]["price"].median())

print("Oil B")
print("Mean:", df[df["item"] == "Oil B"]["price"].mean())
print("Median:", df[df["item"] == "Oil B"]["price"].median())

print("Pulse C")
print("Mean:", df[df["item"] == "Pulse C"]["price"].mean())
print("Median:", df[df["item"] == "Pulse C"]["price"].median())

print("Sweetener D")
print("Mean:", df[df["item"] == "Sweetener D"]["price"].mean())
print("Median:", df[df["item"] == "Sweetener D"]["price"].median())
```

## 3. First-to-last percentage change

I first found the average price of each item on the first date and the last date. Then I used the percentage change formula:

**((Last average price - First average price) / First average price) × 100**

```python
first_date = df["date"].min()
last_date = df["date"].max()

print("First date:", first_date)
print("Last date:", last_date)

# Staple A
first_price = df.loc[(df["item"] == "Staple A") & (df["date"] == first_date), "price"].mean()
last_price = df.loc[(df["item"] == "Staple A") & (df["date"] == last_date), "price"].mean()
change = ((last_price - first_price) / first_price) * 100
print("Staple A percentage change:", change)

# Oil B
first_price = df.loc[(df["item"] == "Oil B") & (df["date"] == first_date), "price"].mean()
last_price = df.loc[(df["item"] == "Oil B") & (df["date"] == last_date), "price"].mean()
change = ((last_price - first_price) / first_price) * 100
print("Oil B percentage change:", change)

# Pulse C
first_price = df.loc[(df["item"] == "Pulse C") & (df["date"] == first_date), "price"].mean()
last_price = df.loc[(df["item"] == "Pulse C") & (df["date"] == last_date), "price"].mean()
change = ((last_price - first_price) / first_price) * 100
print("Pulse C percentage change:", change)

# Sweetener D
first_price = df.loc[(df["item"] == "Sweetener D") & (df["date"] == first_date), "price"].mean()
last_price = df.loc[(df["item"] == "Sweetener D") & (df["date"] == last_date), "price"].mean()
change = ((last_price - first_price) / first_price) * 100
print("Sweetener D percentage change:", change)
```

## 4. Mean and median with and without a flagged value

For this part, I used the Week 2 cleaned file. I selected Sweetener D because it has a flagged value of 9,999. I compared the calculations with this value included and then excluded.

The flagged value is excluded only for this comparison. It is not being deleted from the cleaned file.

```python
week2 = pd.read_csv("teaching_prices_cleaned.csv")

print("Week 2 cleaned file loaded.")
print("Number of rows and columns:", week2.shape)

display(week2.head())
```

```python
sweetener = week2[week2["item"] == "Sweetener D"]

print("With the flagged value included")
print("Mean:", sweetener["price"].mean())
print("Median:", sweetener["price"].median())

sweetener_without_flag = sweetener[sweetener["price"] != 9999]

print("Without the flagged value")
print("Mean:", sweetener_without_flag["price"].mean())
print("Median:", sweetener_without_flag["price"].median())
```

## 5. Short explanation

The flagged value of 9,999 increases the mean because the mean uses all the values in the calculation. The median changes only slightly because it depends on the middle position of the values. This shows that the median is less affected by an unusually high value.

The value is kept in the dataset because being unusual does not prove that it is incorrect.
